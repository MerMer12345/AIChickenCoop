```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your two files
df1 = pd.read_csv('Training/object_countsV8.csv')
df2 = pd.read_csv('Training/object_countsV12.csv')

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(df1['Time (s)'], df1['Object Count'], label='YOLOv8', marker='o')
plt.plot(df2['Time (s)'], df2['Object Count'], label='YOLOv12', marker='x')

plt.title('Object Count Over Time Comparison')
plt.xlabel('Time (s)')
plt.ylabel('Object Count')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python

# Load CSV files
df1 = pd.read_csv('Training/object_countsV8.csv')
df2 = pd.read_csv('Training/object_countsV12.csv')

# Function to compute and round average object count per nearest second
def average_per_rounded_second(df):
    df['Rounded Second'] = df['Time (s)'].round().astype(int)
    avg_df = df.groupby('Rounded Second')['Object Count'].mean().round().astype(int).reset_index()
    return avg_df

avg1 = average_per_rounded_second(df1)
avg2 = average_per_rounded_second(df2)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(avg1['Rounded Second'], avg1['Object Count'], label='YOLOv8', marker='o')
plt.plot(avg2['Rounded Second'], avg2['Object Count'], label='YOLOv12', marker='x')

plt.title('Rounded Average Object Count Per Second')
plt.xlabel('Rounded Second')
plt.ylabel('Rounded Average Object Count')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file with semicolon separator
data = pd.read_csv('ChickenStatistics.csv', delimiter=';')

# Sort data by Year
data = data.sort_values(by='Year')

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(data['Year'], data['Percentage'], marker='o')
plt.title('Percentage over Years')
plt.xlabel('Year')
plt.ylabel('Percentage (%)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
