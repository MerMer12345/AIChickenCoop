```python

from ultralytics import YOLO

# List of model types you want to train
models = ['yolov8n', 'yolo12n', 'yolov5n']

# Path to your dataset YAML
data_yaml = '../ImgLabelling/datasetChickens/data.yaml'

# Output folder to store all models
output_dir = 'trained_models'


# Train each model
for model_type in models:
    print(f"Training {model_type}...")
    model = YOLO(f'{model_type}.pt')  # Load pretrained model
    model.train(
        data=data_yaml,
        epochs=50,
        imgsz=640,
        project=output_dir,
        name=model_type,
        exist_ok=True
    )
    print(f"{model_type} training complete.\n")
```

```python
from ultralytics import YOLO

# Load a previously trained model
model = YOLO('trained_models/yolov8n/weights/best.pt')

# Recognise on an Image that it has never seen
results = model('test_difficult.png')

# Show results
results[0].show()
```
