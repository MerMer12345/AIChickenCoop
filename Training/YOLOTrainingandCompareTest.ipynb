```python
from ultralytics import YOLO
import pandas as pd
import os

def train_model(version: str, model_name: str, data_path: str, epochs: int = 50):
    print(f"🔧 Training {version} on {data_path}")
    model = YOLO(model_name)  # 'yolov5s.pt' or 'yolov8s.pt'
    results = model.train(
        data=data_path,
        epochs=epochs,
        imgsz=640,
        batch=16,
        name=version,
        project="runs_compare",
    )
    return model

def evaluate_model(model: YOLO):
    metrics = model.val()  # returns a Results object with metrics
    return {
        "mAP50": metrics.box.map50,
        "mAP50-95": metrics.box.map,
        "Precision": metrics.precision,
        "Recall": metrics.box.recall
    }

def compare_models():
    results_list = []
    
    versions = {
        "YOLOv5s": "yolov5s.pt",
        "YOLOv8s": "yolov8s.pt"
    }

    for name, model_file in versions.items():
        model = train_model(name, model_file, "../ImgLabelling/datasetChickens/data.yaml")
        metrics = evaluate_model(model)
        results_list.append({"Model": name, **metrics})

    df = pd.DataFrame(results_list)
    print("\n📊 Comparison Results:\n")
    print(df)

if __name__ == "__main__":
    compare_models()
```
