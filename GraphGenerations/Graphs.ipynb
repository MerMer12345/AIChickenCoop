```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your two files
df1 = pd.read_csv('../Detecting/Logs/object_countsV8.csv')
df2 = pd.read_csv('../Detecting/Logs/object_countsV12.csv')

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(df1['Time (s)'], df1['Object Count'], label='YOLOv8', marker='o')
plt.plot(df2['Time (s)'], df2['Object Count'], label='YOLOv12', marker='x')

plt.title('Object Count Over Time Comparison')
plt.xlabel('Time (s)')
plt.ylabel('Object Count')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python

# Load CSV files
df1 = pd.read_csv('../Detecting/Logs/object_countsV8.csv')
df2 = pd.read_csv('../Detecting/Logs/object_countsV12.csv')

# Function to compute and round average object count per nearest second
def average_per_rounded_second(df):
    df['Rounded Second'] = df['Time (s)'].round().astype(int)
    avg_df = df.groupby('Rounded Second')['Object Count'].mean().round().astype(int).reset_index()
    return avg_df

avg1 = average_per_rounded_second(df1)
avg2 = average_per_rounded_second(df2)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(avg1['Rounded Second'], avg1['Object Count'], label='YOLOv8', marker='o')
plt.plot(avg2['Rounded Second'], avg2['Object Count'], label='YOLOv12', marker='x')

plt.title('Rounded Average Object Count Per Second')
plt.xlabel('Rounded Second')
plt.ylabel('Rounded Average Object Count')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
# Load CSV files
df1 = pd.read_csv('../Detecting/Logs/object_countsV8.csv')
df2 = pd.read_csv('../Detecting/Logs/object_countsV12.csv')

# Function to compute and round average object count per 5-second bins
def average_per_five_seconds(df):
    df['Rounded Second'] = (df['Time (s)'] // 5 * 5).astype(int)  # Floor to nearest multiple of 5
    avg_df = df.groupby('Rounded Second')['Object Count'].mean().round().astype(int).reset_index()
    return avg_df

# Apply the updated function
avg1 = average_per_five_seconds(df1)
avg2 = average_per_five_seconds(df2)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(avg1['Rounded Second'], avg1['Object Count'], label='YOLOv8', marker='o')
plt.plot(avg2['Rounded Second'], avg2['Object Count'], label='YOLOv12', marker='x')

plt.title('Rounded Average Object Count Every 5 Seconds')
plt.xlabel('Time (s) (Grouped by 5 seconds)')
plt.ylabel('Average Object Count')
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()

```
